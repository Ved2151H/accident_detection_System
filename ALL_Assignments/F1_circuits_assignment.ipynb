Q1  What is the average latitude of all circuits in the dataset?

```python
import pandas as pd
df = pd.read_csv('circuits.csv')
avg_lat = df['lat'].mean()
print (avg_lat)
```

Q4. What is the median latitude for circuits located in Europe

```python
europe = [
    'UK','United Kingdom','France','Germany','Italy','Spain','Austria','Hungary','Netherlands','Belgium','Portugal',
    'Russia','Finland','Turkey','Greece','Sweden','Denmark','Switzerland','Czech Republic','Slovakia','Poland'
]

median = df[df["country"].isin(europe)]["lat"].median()
print(median)
```
