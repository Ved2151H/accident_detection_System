```python
import string 

stringg = "This is a sample string that we will use for tokenization. Tokenization is the process of breaking down a string into smaller pieces, called tokens. These tokens can be words, phrases, or even characters, depending on the level of tokenization we want to achieve. In this example, we will tokenize the string into individual words."
punct = ".,;:!?()[]{}\"'"

for p in punct:
    stringg = stringg.replace(p, "")
    
tokens = stringg.lower().split()
print(tokens)
```

```python
tokens = []
start = 0 

for word in stringg.split():
    start = stringg.index(word, start  )
    end = start + len(word)
    tokens.append((word, start, end))
    start = end
    
print(tokens)
```
