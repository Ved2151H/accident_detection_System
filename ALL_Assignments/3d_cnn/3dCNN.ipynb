```python
import torch
import torch.nn as nn
from PIL import Image
import torchvision.transforms as T


img = Image.open("img.jpg")
img = T.ToTensor()(img)
img = img.unsqueeze(0)
img = img.unsqueeze(2)


model = nn.Sequential(
    nn.Conv3d(3, 8, 3, padding=1),
    nn.ReLU(),
    nn.MaxPool3d((1,2,2)),  
    nn.Conv3d(8, 16, 3, padding=1),
    nn.ReLU(),
    nn.AdaptiveAvgPool3d(1)
)


x = img
for layer in model:
    x = layer(x)
    print("\nlaayer:", layer.__class__.__name__)
    print("shape:", x.shape)
    print("mean:", x.mean().item()) 
    


gap = x.view(x.size(0), -1)

print("\n globel avg")
print(gap)
```
