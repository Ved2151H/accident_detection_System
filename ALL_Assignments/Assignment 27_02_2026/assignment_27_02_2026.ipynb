## 2. Define a Perceptron. Write its mathematical equation and explain each component of the equation.

# Perceptron

## Definition

A **Perceptron** is a supervised learning algorithm used for **binary classification**.  
It is the simplest type of artificial neural network and models a single artificial neuron.  
It learns a **linear decision boundary** that separates data into two classes.

The perceptron works only when the dataset is **linearly separable**.

---

## Mathematical Equation

$$
y = f\left(\sum_{i=1}^{n} w_i x_i + b \right)
$$

Vector form:

$$
y = f(\mathbf{w} \cdot \mathbf{x} + b)
$$

---

## Explanation of Components

### 1. Input Vector (\(\mathbf{x}\))

$$
\mathbf{x} = (x_1, x_2, \dots, x_n)
$$

- Represents the input features  
- Each \(x_i\) is a numerical feature value  

---

### 2. Weight Vector (\(\mathbf{w}\))

$$
\mathbf{w} = (w_1, w_2, \dots, w_n)
$$

- Each weight corresponds to one input feature  
- Determines the importance of that feature  
- Larger weight → stronger influence  

---

### 3. Bias (\(b\))

- A scalar constant  
- Shifts the decision boundary  
- Allows the model to classify data that does not pass through the origin  

---

### 4. Weighted Sum

$$
z = \sum_{i=1}^{n} w_i x_i + b
$$

This is the linear combination of inputs and weights.

---

### 5. Activation Function

The perceptron uses a **step function**:

$$
f(z) =
\begin{cases}
1 & \text{if } z \ge 0 \\
0 & \text{if } z < 0
\end{cases}
$$

- Produces binary output  
- Converts continuous value into class label  

---

## Decision Boundary

$$
w_1x_1 + w_2x_2 + \dots + w_nx_n + b = 0
$$

- In 2D → Straight line  
- In 3D → Plane  
- In higher dimensions → Hyperplane  

---

## Key Characteristics

- Used for binary classification  
- Single-layer neural network  
- No hidden layers  
- Works only for linearly separable data  
- Simple and computationally efficient  

---

```python
import numpy as np

class Perceptron:
    def __init__(self, learning_rate=0.1, epochs=10):
        self.lr = learning_rate
        self.epochs = epochs
    
    def step_function(self, z):
        return np.where(z >= 0, 1, 0)
    
    def fit(self, X, y):
        n_samples, n_features = X.shape
        self.weights = np.zeros(n_features)
        self.bias = 0
        
        for _ in range(self.epochs):
            for idx, x_i in enumerate(X):
                linear_output = np.dot(x_i, self.weights) + self.bias
                y_pred = self.step_function(linear_output)
                
                update = self.lr * (y[idx] - y_pred)
                self.weights += update * x_i
                self.bias += update
    
    def predict(self, X):
        linear_output = np.dot(X, self.weights) + self.bias
        return self.step_function(linear_output)


# Example Dataset
X = np.array([[2, 3],
              [1, 1],
              [2, 1],
              [3, 2]])

y = np.array([1, 0, 0, 1])

# Train Model
model = Perceptron(learning_rate=0.1, epochs=20)
model.fit(X, y)

# Predictions
predictions = model.predict(X)

print("Weights:", model.weights)
print("Bias:", model.bias)
print("Predictions:", predictions)
```

## 3. What is an activation function? Explain the purpose of activation functions and describe Sigmoid,ReLU, and Tanh with their advantages and disadvantages

# Activation Function

## Definition

An **activation function** is a mathematical function applied to the output of a neuron 
(after the weighted sum and bias) to introduce **non-linearity** into the neural network.

Mathematically:

$$
z = \sum_{i=1}^{n} w_i x_i + b
$$

$$
y = f(z)
$$

Where:
- \(z\) = weighted sum  
- \(f(z)\) = activation function  
- \(y\) = final output  

---

# Purpose of Activation Functions

Activation functions are essential because:

###  Introduce Non-Linearity
Without activation functions, a neural network becomes just a linear model.
Stacking multiple linear layers still produces a linear function.

###  Enable Complex Learning
They allow the network to learn:
- Non-linear patterns
- Complex relationships
- Image, speech, and text representations

###  Control Output Range
Different activation functions restrict outputs to specific ranges:
- Sigmoid → (0, 1)
- Tanh → (-1, 1)
- ReLU → [0, ∞)

###  Help in Gradient-Based Learning
They influence gradient flow during backpropagation.

---

#  Sigmoid Activation Function

## Formula

$$
\sigma(z) = \frac{1}{1 + e^{-z}}
$$

## Output Range

$$
0 < \sigma(z) < 1
$$

## Characteristics

- S-shaped curve
- Used for binary classification
- Converts output into probability

## Advantages

✔ Output interpretable as probability  
✔ Smooth and differentiable  
✔ Suitable for logistic regression  

## Disadvantages

✘ Vanishing gradient problem  
✘ Not zero-centered  
✘ Slow convergence for deep networks  

---

#  ReLU (Rectified Linear Unit)

## Formula

$$
f(z) = \max(0, z)
$$

## Output Range

$$
[0, \infty)
$$

## Characteristics

- Most widely used in deep learning  
- Computationally efficient  

## Advantages

✔ Reduces vanishing gradient problem  
✔ Faster training  
✔ Simple computation  

## Disadvantages

✘ Dying ReLU problem (neurons may stop updating)  
✘ Not zero-centered  
✘ Output unbounded  

---

#  Tanh (Hyperbolic Tangent)

## Formula

$$
\tanh(z) = \frac{e^z - e^{-z}}{e^z + e^{-z}}
$$

## Output Range

$$
-1 < \tanh(z) < 1
$$

## Characteristics

- Zero-centered  
- Stronger gradients than sigmoid  

## Advantages

✔ Zero-centered output  
✔ Better than sigmoid for hidden layers  
✔ Smooth and differentiable  

## Disadvantages

✘ Still suffers from vanishing gradient  
✘ Computationally heavier than ReLU  

---

# Comparison Table

| Function | Formula | Range | Main Advantage | Main Disadvantage |
|-----------|----------|--------|----------------|------------------|
| Sigmoid | \( \frac{1}{1+e^{-z}} \) | (0,1) | Probability output | Vanishing gradient |
| ReLU | \( \max(0,z) \) | [0,∞) | Fast & efficient | Dying ReLU |
| Tanh | \( \frac{e^z - e^{-z}}{e^z + e^{-z}} \) | (-1,1) | Zero-centered | Vanishing gradient |

---



```python
import numpy as np
import matplotlib.pyplot as plt

# Create input values
x = np.linspace(-10, 10, 100)

# 1️⃣ Sigmoid Function
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

y_sigmoid = sigmoid(x)

plt.figure()
plt.plot(x, y_sigmoid)
plt.title("Sigmoid Activation Function")
plt.xlabel("Input")
plt.ylabel("Output")
plt.grid(True)
plt.show()


# 2️⃣ ReLU Function
def relu(z):
    return np.maximum(0, z)

y_relu = relu(x)

plt.figure()
plt.plot(x, y_relu)
plt.title("ReLU Activation Function")
plt.xlabel("Input")
plt.ylabel("Output")
plt.grid(True)
plt.show()


# 3️⃣ Tanh Function
def tanh(z):
    return np.tanh(z)

y_tanh = tanh(x)

plt.figure()
plt.plot(x, y_tanh)
plt.title("Tanh Activation Function")
plt.xlabel("Input")
plt.ylabel("Output")
plt.grid(True)
plt.show()
```

## 4. Explain the architecture of a basic Artificial Neural Network (ANN). Describe input layer, hidden layer, output layer, weights, bias, and forward propagation.

# Architecture of a Basic Artificial Neural Network (ANN)

## Definition

An **Artificial Neural Network (ANN)** is a computational model inspired by the human brain.  
It consists of interconnected neurons arranged in layers that process information and learn patterns from data.

A basic ANN contains:
- Input Layer  
- Hidden Layer(s)  
- Output Layer  
- Weights  
- Bias  
- Forward Propagation mechanism  

---

# Structure of a Basic ANN

Input → Hidden Layer(s) → Output  

---

##  Input Layer

The **input layer** receives raw data from the dataset.

If there are \(n\) features in the dataset, the input layer contains \(n\) neurons.

$$
\mathbf{x} = (x_1, x_2, \dots, x_n)
$$

- Each neuron represents one feature  
- No computation is performed here  
- It only forwards data to the next layer  

---

##  Hidden Layer

The **hidden layer** performs computations and extracts patterns.

Each neuron computes:

$$
z = \sum_{i=1}^{n} w_i x_i + b
$$

Then applies activation:

$$
a = f(z)
$$

Where:
- \(w_i\) = weights  
- \(b\) = bias  
- \(f(z)\) = activation function  

Hidden layers introduce **non-linearity** and allow the network to learn complex patterns.

---

##  Output Layer

The **output layer** produces the final prediction.

Output is calculated as:

$$
y = f\left(\sum w_i a_i + b\right)
$$

Number of output neurons depends on the problem:

- Binary classification → 1 neuron  
- Multi-class classification → number of classes  
- Regression → 1 neuron  

---

##  Weights

Weights determine the strength of connections between neurons.

- Each connection has a weight  
- Large weight → strong influence  
- Weights are updated during training  

---

##  Bias

Bias is an additional parameter added to each neuron.

- Helps shift the decision boundary  
- Increases flexibility of the model  
- Prevents forcing output through origin  

---

##  Forward Propagation

Forward propagation is the process of passing input data through the network to generate output.

### Step 1: Hidden Layer Computation

$$
z^{(1)} = W^{(1)}x + b^{(1)}
$$

$$
a^{(1)} = f(z^{(1)})
$$

### Step 2: Output Layer Computation

$$
z^{(2)} = W^{(2)}a^{(1)} + b^{(2)}
$$

$$
y = f(z^{(2)})
$$

Flow of data:

Input → Weighted Sum → Activation → Output  

---

# Key Characteristics

- Feedforward structure  
- Fully connected layers  
- Uses activation functions  
- Learns by adjusting weights and bias  
- Can model non-linear relationships  

---



```python
import numpy as np

def relu(z):
    return np.maximum(0, z)

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

class BasicANN:
    def __init__(self, input_size, hidden_size, output_size):
        self.W1 = np.random.randn(input_size, hidden_size)
        self.b1 = np.zeros((1, hidden_size))
        
        self.W2 = np.random.randn(hidden_size, output_size)
        self.b2 = np.zeros((1, output_size))
    
    def forward(self, X):
        self.z1 = np.dot(X, self.W1) + self.b1
        self.a1 = relu(self.z1)
        
        self.z2 = np.dot(self.a1, self.W2) + self.b2
        self.output = sigmoid(self.z2)
        
        return self.output

X = np.array([
    [1, 2, 3],
    [2, 1, 0],
    [3, 2, 1],
    [0, 1, 2]
])

model = BasicANN(input_size=3, hidden_size=4, output_size=1)
output = model.forward(X)

print("Output of ANN:")
print(output)
```

## 5. What is backpropagation? Explain the role of the chain rule in updating weights. Why is gradient descent used?

# Backpropagation

## Definition

**Backpropagation** is an algorithm used to train artificial neural networks by 
updating weights and bias values to minimize the error between predicted output and actual output.

It works by computing the gradient of the loss function with respect to each weight 
using the **chain rule of calculus**, and then updating the weights using **gradient descent**.

---

# Why Backpropagation is Needed

During forward propagation:

$$
z = Wx + b
$$

$$
a = f(z)
$$

We get a prediction \( \hat{y} \).

The error (loss) is:

$$
L = \frac{1}{2}(y - \hat{y})^2
$$

To reduce this loss, we must update the weights in the direction that reduces error.

---

# Role of the Chain Rule

The loss depends on weights indirectly:

$$
L \rightarrow \hat{y} \rightarrow z \rightarrow w
$$

To compute how loss changes with weight:

$$
\frac{\partial L}{\partial w}
=
\frac{\partial L}{\partial \hat{y}}
\cdot
\frac{\partial \hat{y}}{\partial z}
\cdot
\frac{\partial z}{\partial w}
$$

This is the **chain rule**.

It allows us to compute gradients layer by layer from output to input.

---

# Gradient Descent

After computing gradient:

$$
w_{new} = w_{old} - \eta \frac{\partial L}{\partial w}
$$

Where:
- \( \eta \) = learning rate
- \( \frac{\partial L}{\partial w} \) = gradient

Gradient descent moves weights in the direction that reduces loss.

---

# Why Gradient Descent is Used

 It minimizes the loss function  
 It is computationally efficient  
 Works well for large neural networks  
 Supports mini-batch and stochastic updates  

---

# Backpropagation Steps

1. Forward propagation  
2. Compute loss  
3. Compute gradients using chain rule  
4. Update weights using gradient descent  
5. Repeat until convergence  

---



```python
import numpy as np

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def sigmoid_derivative(z):
    return sigmoid(z) * (1 - sigmoid(z))

X = np.array([[0], [1], [2], [3]])
y = np.array([[0], [0], [1], [1]])

w = np.random.randn(1, 1)
b = np.zeros((1, 1))

learning_rate = 0.1
epochs = 1000

for epoch in range(epochs):
    z = np.dot(X, w) + b
    y_pred = sigmoid(z)

    loss = np.mean((y - y_pred) ** 2)

    dL_dy = 2 * (y_pred - y) / len(y)
    dy_dz = sigmoid_derivative(z)
    dz_dw = X

    dL_dw = np.dot(dz_dw.T, dL_dy * dy_dz)
    dL_db = np.sum(dL_dy * dy_dz)

    w -= learning_rate * dL_dw
    b -= learning_rate * dL_db

print("Trained Weight:", w)
print("Trained Bias:", b)
print("Final Predictions:")
print(sigmoid(np.dot(X, w) + b))
```

## 6. Differentiate between Batch Gradient Descent, Stochastic Gradient Descent (SGD), and Mini-batch Gradient Descent.

# Difference Between Batch Gradient Descent, Stochastic Gradient Descent (SGD), and Mini-batch Gradient Descent

## Introduction

Gradient Descent is an optimization algorithm used to minimize the loss function 
by updating weights in the direction of the negative gradient.

General update rule:

$$
w_{new} = w_{old} - \eta \frac{\partial L}{\partial w}
$$

Where:
- \( \eta \) = learning rate  
- \( \frac{\partial L}{\partial w} \) = gradient of loss  

The difference between Batch, SGD, and Mini-batch lies in **how much data is used to compute the gradient**.

---

#  Batch Gradient Descent

## Definition

Batch Gradient Descent computes the gradient using the **entire training dataset**.

If dataset has \(N\) samples:

$$
\frac{\partial L}{\partial w} = \frac{1}{N} \sum_{i=1}^{N} \frac{\partial L_i}{\partial w}
$$

## Characteristics

- Uses all data at once  
- One update per epoch  
- Stable convergence  

## Advantages

✔ Smooth and stable updates  
✔ Guaranteed convergence for convex problems  

## Disadvantages

✘ Slow for large datasets  
✘ High memory usage  
✘ Cannot handle very large datasets efficiently  

---

#  Stochastic Gradient Descent (SGD)

## Definition

Stochastic Gradient Descent updates weights using **one training example at a time**.

$$
\frac{\partial L}{\partial w} = \frac{\partial L_i}{\partial w}
$$

## Characteristics

- Updates weights after every sample  
- Faster updates  
- Noisy convergence  

## Advantages

✔ Faster for large datasets  
✔ Can escape local minima  
✔ Requires less memory  

## Disadvantages

✘ High variance (noisy updates)  
✘ May fluctuate around minimum  
✘ Less stable convergence  

---

#  Mini-batch Gradient Descent

## Definition

Mini-batch Gradient Descent computes gradient using a **small subset (batch) of data**.

If batch size = \(m\):

$$
\frac{\partial L}{\partial w} = \frac{1}{m} \sum_{i=1}^{m} \frac{\partial L_i}{\partial w}
$$

## Characteristics

- Compromise between Batch and SGD  
- Common batch sizes: 16, 32, 64, 128  
- Most widely used in deep learning  

## Advantages

✔ Faster than Batch  
✔ More stable than SGD  
✔ Efficient GPU utilization  

## Disadvantages

✘ Requires tuning batch size  
✘ Still has some noise  

---

# Comparison Table

| Feature | Batch GD | SGD | Mini-batch GD |
|----------|-----------|------|----------------|
| Data Used | Entire dataset | 1 sample | Small subset |
| Speed | Slow | Fast | Moderate |
| Stability | Very stable | Noisy | Balanced |
| Memory Usage | High | Low | Moderate |
| Convergence | Smooth | Fluctuating | Smooth with small noise |
| GPU Efficiency | Poor | Poor | Excellent |
| Best For | Small datasets | Very large datasets | Deep learning models |

---

# Key Insight

- Batch GD → Accurate but slow  
- SGD → Fast but noisy  
- Mini-batch GD → Best practical choice  

Most modern neural networks use **Mini-batch Gradient Descent**.

---



```python
import numpy as np


X = np.array([[1], [2], [3], [4], [5]])
y = np.array([[2], [4], [6], [8], [10]])   


def initialize():
    w = np.random.randn(1, 1)
    b = np.zeros((1, 1))
    return w, b


def compute_loss(X, y, w, b):
    y_pred = X @ w + b
    return np.mean((y - y_pred) ** 2)


def batch_gradient_descent(X, y, lr=0.01, epochs=50):
    w, b = initialize()
    n = len(X)

    for _ in range(epochs):
        y_pred = X @ w + b
        
        dw = (-2/n) * X.T @ (y - y_pred)
        db = (-2/n) * np.sum(y - y_pred)

        w -= lr * dw
        b -= lr * db

    return w, b


def stochastic_gradient_descent(X, y, lr=0.01, epochs=50):
    w, b = initialize()
    n = len(X)

    for _ in range(epochs):
        for i in range(n):
            xi = X[i].reshape(1, -1)
            yi = y[i].reshape(1, -1)

            y_pred = xi @ w + b

            dw = -2 * xi.T @ (yi - y_pred)
            db = -2 * np.sum(yi - y_pred)

            w -= lr * dw
            b -= lr * db

    return w, b


def mini_batch_gradient_descent(X, y, lr=0.01, epochs=50, batch_size=2):
    w, b = initialize()
    n = len(X)

    for _ in range(epochs):
        indices = np.random.permutation(n)
        X_shuffled = X[indices]
        y_shuffled = y[indices]

        for i in range(0, n, batch_size):
            X_batch = X_shuffled[i:i+batch_size]
            y_batch = y_shuffled[i:i+batch_size]

            m = len(X_batch)
            y_pred = X_batch @ w + b

            dw = (-2/m) * X_batch.T @ (y_batch - y_pred)
            db = (-2/m) * np.sum(y_batch - y_pred)

            w -= lr * dw
            b -= lr * db

    return w, b


w_batch, b_batch = batch_gradient_descent(X, y)
w_sgd, b_sgd = stochastic_gradient_descent(X, y)
w_mini, b_mini = mini_batch_gradient_descent(X, y)

print("Batch GD -> w:", w_batch.flatten()[0], " b:", b_batch.flatten()[0])
print("SGD -> w:", w_sgd.flatten()[0], " b:", b_sgd.flatten()[0])
print("Mini-batch GD -> w:", w_mini.flatten()[0], " b:", b_mini.flatten()[0])
```
