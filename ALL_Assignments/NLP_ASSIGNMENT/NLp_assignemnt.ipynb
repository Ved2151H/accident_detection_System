# NLP Preprocessing, NER, and BoW


## Section 1: Preprocessing & Text Cleaning

```python
import re

text = "I love this 😊!!!"
cleaned = re.sub(r'[^\w\s]', '', text)
print("Original:", text)
print("Cleaned:", cleaned)  
```

```python
# US phone number regex
import re

text = "Call me at (555) 123-4567 or +44 20 7123 1234"
pattern = r'\(\d{3}\)\s\d{3}-\d{4}'
matches = re.findall(pattern, text)
print(matches)  
```

## Section 2: Stemming vs Lemmatization

```python
from nltk.stem import PorterStemmer, WordNetLemmatizer

stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

word = "better"

print("Stem:", stemmer.stem(word))
print("Lemma:", lemmatizer.lemmatize(word, pos='a'))  
```

## Section 3: POS Tagging & NER

```python
import nltk
nltk.download('punkt')
nltk.download('averaged_perceptron_tagger')

sentence = "I will book the flight after I read this book"
tokens = nltk.word_tokenize(sentence)
tags = nltk.pos_tag(tokens)

print(tags)
```

## Section 4: Bag of Words

```python
from sklearn.feature_extraction.text import CountVectorizer

corpus = ["I eat the apple", "I like the apple"]
vectorizer = CountVectorizer(vocabulary=['apple','banana','eat','i','like'])

X = vectorizer.fit_transform(corpus)
print(X.toarray())
```

## Complete Pipeline Example

```python
import nltk
from sklearn.feature_extraction.text import CountVectorizer

text = "I do not like this product!"


tokens = nltk.word_tokenize(text)


pos_tags = nltk.pos_tag(tokens)


stopwords = {'this'}
filtered = [w for w in tokens if w.lower() not in stopwords]


from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
lemmas = [lemmatizer.lemmatize(w.lower()) for w in filtered]


vectorizer = CountVectorizer()
X = vectorizer.fit_transform([" ".join(lemmas)])

print("Final tokens:", lemmas)
print("BoW:", X.toarray())
```
