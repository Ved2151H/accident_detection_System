 Question 1: Clean the Dataset & Find the Most Frequent Category
 You are given a list representing a categorical column from a dataset.
 Some values are invalid (e.g., "?", "N/A", "None", ""). Your task is to clean the data
 using loops and then find the most frequent valid category.
 Input Example:
 data = ["red", "blue", "?", "green", "blue", "", "N/A", "green", "blue"]
 Requirements:
 1. Create a new list that contains only valid categories.
 2. Without using any library function like Counter, manually compute the frequency of each category.
 3. Return the category with the highest frequency and its count.
 Output Example:
 Most Frequent Category: blue (3 times)

```python
list = ["red", "blue", "?", "green", "blue", "", "N/A", "green", "blue"]

for i in range(len(list)):
    if list[i] == "?" or list[i] == "" or list[i] == "N/A":
        i = i + 1
    else:
        print(list[i]  )
```

 Question 2: Compute Moving Average (Window Size K)
 You are given a list of numbers representing daily temperatures.
 Your task is to manually compute the moving average for a given window size k
 (without using NumPy or pandas).
 Input Example:
 temps = [30, 32, 31, 29, 28, 35, 36]
 k = 3
 Requirements:
 1. Use loops to compute the sum of every sliding window of size k.
2. Append each window's average to a list.
 3. Return the list of moving averages.
 Expected Output:
 [31.0, 30.67, 29.33, 30.67, 33.0]

```python
temps =[30, 32, 31, 29, 28, 35, 36]
list = []
for i in range(len(temps)-2):
    a = temps[i] + temps[i+1] + temps[i+2]
    avg = a / 3
    list.append(avg)
    
print(list)
    
```
