# Problem 1: Transaction Frequency Analyzer

```python
T = [200, 500, 200, 100, 500, 200, 100, 500, 500]
freq = {}


for t in T:
    if t in freq:
        freq[t] += 1
    else:
        freq[t] = 1
        

sorted_freq = sorted(freq, key=lambda x: (-freq[x], x))

final = sorted_freq[:2]
print(final)
```
